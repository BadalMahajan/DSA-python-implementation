```python
#Linear Search - based on brute force (comparing every element with the input)
#Don't need a sorted array
#Time complexity is O(n)
```

```python
def LinearSearch(arr,item):
    for i in range(len(arr)):
        if arr[i] == item:
            return i
    return -1
```

```python
arr = [1,2,4,3,5,9,8]
```

```python
LinearSearch(arr,9)
```

```python
# Binary Search
#Only works in sorted array
```

```python
def BinarySearch(arr,low,high,item):
    if low <= high:
        mid = (low + high)//2
        
        if arr[mid] == item:
            return mid
        elif arr[mid] > item:
            return BinarySearch(arr,low,mid-1,item)
        else:
            return BinarySearch(arr,mid+1,high,item)
    else:
        return -1
```

```python
arr = [12,24,35,54,68,80,99,100]
```

```python
BinarySearch(arr,0,len(arr)-1,80)
```
