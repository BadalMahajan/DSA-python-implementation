```python
class Node:
    def __init__(self,key,value):
        self.key = key
        self.value = value
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0
    
    def append(self,key,value):
        new_node = Node(key,value)
        
        if self.head == None:
            self.head = new_node
            self.n += 1
            return
        
        curr = self.head
        while curr.next != None:
            curr = curr.next
            
        curr.next = new_node
        self.n += 1
        
    def clear(self):
        self.head = None
        self.n = 0
    
    def delete_head(self):
        if self.head == None:   # For Empty Linked List
            return "Empty Linked List" 
        self.head = self.head.next
        self.n -= 1
    
    def size(self):
        return self.n
        
    def remove(self,key):
        
        if self.head == None:
            return -1
        
        if self.head.key == key:
            return self.delete_head()
        
        curr = self.head
        
        while curr.next != None:
            if curr.next.key == key:
                break
            curr = curr.next
        if curr.next == None:
            return -1
        else:
            curr.next = curr.next.next
            self.n -= 1
            
    def __delitem__(self,index):
        curr = self.head
        pos = 0
        
        if index == 0:
            return self.delete_head()
        
        while curr.next != None:
            if pos == index -1:
                break
            curr = curr.next
            pos += 1
        if curr.next == None:
            return "Item Not Found"
        else:
            curr.next = curr.next.next
            self.n -= 1
    
    def search(self,key):
        curr = self.head
        pos = 0
        
        while curr != None:
            if curr.key == key:
                return pos
            curr = curr.next
            pos += 1
        return -1
    
    def __getitem__(self, index):
        curr = self.head
        pos = 0
        
        while curr != None:
            if pos == index:
                return curr.data
            curr = curr.next
            pos += 1
        return "IndexError"
    
    def get_node_at_index(self,index):
        temp = self.head
        counter = 0
        
        while temp is not None:
            if counter == index:
                return temp
            counter += 1
            temp = temp.next
    
    def __len__(self):
        return self.n
    
    def traverse(self):
        curr = self.head
        
        while curr != None:
            print(curr.key, "-->", curr.value, end = " ")
            curr = curr.next
            
    def __str__(self):
        curr = self.head
        if curr == None:
            return "Empty Linked List"
        result = ''
        while curr != None:
            result = result + str(curr.key) + '->'
            curr = curr.next
            
        return result[:-2]
```

```python
class Dictionary:
    
    def __init__(self,capacity):
        self.capacity = capacity
        self.size = 0
        # create an array where each element is a Linkedlist or (buckets)
        self.buckets = self.make_array_of_LL(capacity)
        
        
    def __getitem__(self,key):
        return self.get(key)
        
    def get(self,key):
        
        bucket_index = self.hash_function(key)
        
        node_index = self.buckets[bucket_index].search(key)
        
        if node_index == -1:
            return "Not Found"
        else:
            node = self.buckets[bucket_index].get_node_at_index(node_index)
            return node.value
        
    def __setitem__(self,key,value):
        self.put(key,value)
        
    def put(self,key,value):
        bucket_index = self.hash_function(key)
        
        node_index = self.get_node_at_index(bucket_index,key)
        
        if node_index == -1:
            #insert
            self.buckets[bucket_index].append(key,value)
            self.size+=1
            
            load_factor = self.size/self.capacity
            print(load_factor)
            
            if (load_factor >= 2):
                self.rehash()
            
        else:
            #update
            node = self.buckets[bucket_index].get_node_at_index(node_index)
            node.value = value
    
    def delete(self,key):
        bucket_index = self.hash_function(key)
        
        res = self.buckets[bucket_index].remove(key)
        
        if res == -1:
            return "Not Found"
        else:
            self.size -=1
            return res
        
    def __delitem__(self,key):
        self.delete(key)
        
    def __str__(self):
        
        for i in self.buckets:
            print(i.traverse(), end = "|")
            
        return ""
    
    def rehash(self):
        
        self.capacity = self.capacity * 2
        old_buckets = self.buckets
        self.size = 0
        self.buckets = self.make_array_of_LL(self.capacity)
        
        for i in old_buckets:
            for j in range(i.size()):
                node = i.get_node_at_index(j)
                key_item = node.key
                value_item = node.value
                self.put(key_item,value_item)
                
    def get_node_at_index(self,bucket_index,key):
        
        node_index = self.buckets[bucket_index].search(key)
        return node_index
        
        
    def make_array_of_LL(self, capacity):
#         L = [LinkedList()] * 3  
#         this will create a same address for every linkedlist which ends up apply operations on all element(LL) of the array
        
        L = []
        for i in range(capacity):
            L.append(LinkedList())
            
        return L
    
    def hash_function(self,key):
        return abs(hash(key)) % self.capacity
```
