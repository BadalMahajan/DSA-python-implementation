```python
class Node:
    def __init__(self,value):
        self.value = value
        self.prev = None
        self.next = None
```

```python
class DoubleLinkedList:
    
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0
        
    def add(self,value):
        
        new_node = Node(value)
        
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            self.size +=1
            
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
            self.size +=1
        
        return new_node.value
            
    def __len__(self):
        return self.size
    
    def traverse(self):
        
        curr = self.head
        
        while curr != None:
            print(curr.value,end= "-->")
            curr = curr.next
            
    def __str__(self):
        
        curr = self.head
        result = ""
        
        while curr != None:
            result = result + str(curr.value) + "-->"
            curr = curr.next
            
        return result[:-3]
    
    def search_index(self,value):
        pos = 0
        curr = self.head
        
        while curr != None:
            if curr.value == value:
                return pos
            curr = curr.next
            pos +=1
        return "Not found"
    
    def __getitem__(self,value):
        return self.search_index(value)
    
    def search_node(self,value):
        curr = self.head
        
        while curr != None:
            if curr.value == value:
                return curr
            curr = curr.next
            
        return None 
    
    def remove(self,value):
        
        node = self.search_node(value)
        
        if node is not None:
            if node.prev is None:
                self.head = node.next
            else:
                node.prev.next = node.next

            if node.next is None:
                self.tail = node.prev
            else:
                node.next.prev = node.prev

            self.size -= 1

        else:
            return "Value not found"
        
    def __delitem__(self,value):
        self.remove(value)
            
            
```
